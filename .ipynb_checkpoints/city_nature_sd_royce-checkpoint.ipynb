# Geospatial Analysis

```python
real_sd_zips = gpd.read_file("zip_data/ZIP_CODES_20260128.geojson")
```

```python
real_sd_zips
```

```python
real_sd_zips = real_sd_zips.to_crs('EPSG:4326')

geometry = [Point(xy) for xy in zip(df['longitude'], df['latitude'])]

observations_gdf = gpd.GeoDataFrame(df, geometry=geometry, crs='EPSG:4326')

obs_with_zip = gpd.sjoin(observations_gdf, real_sd_zips, #joins the two dataframes on the geographic location paired with zipcode
                          how='left', predicate='within')

zip_counts = obs_with_zip.groupby('zip').size().reset_index(name='observation_count')

zip_map = real_sd_zips.merge(zip_counts, 
                                on = 'zip',
                                how='left')
```

```python
zip_map['observation_count'] = zip_map['observation_count'].fillna(0)
zip_map
```

## plotly ver.

```python
sd_zip_map = zip_map
fig = px.choropleth_mapbox(
    sd_zip_map,
    geojson=sd_zip_map.geometry,
    locations=sd_zip_map.index,
    color='observation_count',
    color_continuous_scale='YlOrRd',
    mapbox_style='open-street-map',
    zoom=9,
    center={'lat': 32.7157, 'lon': -117.1611},
    opacity=0.7,
    hover_data={
        'zip': True,
        'community': True,  
        'observation_count': True
    }
)

fig.update_traces(
    hovertemplate='<b>ZIP: %{customdata[0]}</b><br>' +
                  'Community: %{customdata[1]}<br>' +
                  'Observations: %{customdata[2]}<br>' +
                  '<extra></extra>'
)

fig.update_layout(
    title='City Nature Challenge 2025: San Diego County<br>Total Observations by ZIP Code',
    height=700,
    margin={"r":0,"t":50,"l":0,"b":0}
)

fig.show()
```

```python
def integrateZip(df, real_sd_zips):
    
    geometry = [Point(xy) for xy in zip(df['longitude'], df['latitude'])]
    
    observations_gdf = gpd.GeoDataFrame(df, geometry=geometry, crs='EPSG:4326')
    
    obs_with_zip = gpd.sjoin(observations_gdf, real_sd_zips, #joins the two dataframes on the geographic location paired with zipcode
                              how='left', predicate='within')
    return obs_with_zip
```

```python


ts_col = "created_at" if "created_at" in df.columns else ("updated_at" if "updated_at" in df.columns else None)
if ts_col is None:
    raise ValueError("Need created_at or updated_at for retention analysis.")

df[ts_col] = pd.to_datetime(df[ts_col], errors="coerce", utc=True)
df = df.dropna(subset=[ts_col, "user_id"])

df["activity_day"] = df[ts_col].dt.floor("D")
active_days = df.groupby("user_id")["activity_day"].nunique()

def bucket_days(n: int) -> str:
    if n <= 1: return "1 day"
    if n == 2: return "2 days"
    if n == 3: return "3 days"
    return "4+ days"

bucket = active_days.apply(bucket_days)

#Observations per user
obs_per_user = df.groupby("user_id").size()

# Aggregate per bucket
order = ["1 day", "2 days", "3 days", "4+ days"]

users_per_bucket = bucket.value_counts().reindex(order, fill_value=0)

avg_obs_per_bucket = (
    pd.DataFrame({"bucket": bucket, "obs": obs_per_user})
    .groupby("bucket")["obs"]
    .mean()
    .reindex(order)
)

# 6) Two separate graphs side by side
fig, axes = plt.subplots(1, 2, figsize=(12, 4), sharex=True)

# Left plot: # users
axes[0].bar(order, users_per_bucket.values)
axes[0].set_title("# of users by active-day bucket")
axes[0].set_ylabel("# users")
axes[0].set_xlabel("Active days (bucket)")
axes[0].tick_params(axis="x", rotation=20)

# Right plot: avg obs/user
axes[1].bar(order, avg_obs_per_bucket.values)
axes[1].set_title("Avg observations per user by bucket")
axes[1].set_ylabel("Avg obs/user")
axes[1].set_xlabel("Active days (bucket)")
axes[1].tick_params(axis="x", rotation=20)

plt.tight_layout()
plt.show()

# Optional: show the table
summary = pd.DataFrame({
    "users": users_per_bucket,
    "avg_obs_per_user": avg_obs_per_bucket
})
print(summary)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = combined_df.copy()

# Pick your species key (taxon_id is best; fall back to scientific_name)
species_col = "taxon_id" if "taxon_id" in df.columns else "scientific_name"
date_col = "observed_on" if "observed_on" in df.columns else "time_observed_at"

df[date_col] = pd.to_datetime(df[date_col], errors="coerce", utc=True)
df = df.dropna(subset=[date_col, species_col])

df["month"] = df[date_col].dt.month

# Define seasons (edit if you want: spring=Mar–May is typical)
spring_months = [3, 4, 5]
post_months   = [6, 7, 8, 9, 10, 11, 12]

spring_species = set(df.loc[df["month"].isin(spring_months), species_col].unique())
post_species   = set(df.loc[df["month"].isin(post_months),   species_col].unique())

overlap = spring_species & post_species
union   = spring_species | post_species

pct_post_also_spring = (len(overlap) / len(post_species)) if len(post_species) else np.nan
jaccard = (len(overlap) / len(union)) if len(union) else np.nan

print(f"Spring species: {len(spring_species)}")
print(f"Post-spring species: {len(post_species)}")
print(f"Overlap: {len(overlap)}")
print(f"% of post-spring species also seen in spring: {pct_post_also_spring:.3f}")
print(f"Jaccard similarity (spring vs post): {jaccard:.3f}")

# Quick bar summary
cats = ["Spring only", "Both", "Post only"]
vals = [len(spring_species - post_species), len(overlap), len(post_species - spring_species)]

plt.figure()
plt.bar(cats, vals)
plt.ylabel("# species")
plt.title("Species overlap: Spring vs Post-spring")
plt.tight_layout()
plt.show()
```

```python
def jaccard_set(a, b):
    a, b = set(a), set(b)
    u = a | b
    return (len(a & b) / len(u)) if len(u) else np.nan

spring_set = spring_species

months = list(range(1, 13))
jac_by_month = []
n_by_month = []

for m in months:
    m_set = set(df.loc[df["month"] == m, species_col].unique())
    jac_by_month.append(jaccard_set(spring_set, m_set))
    n_by_month.append(len(m_set))

plt.figure()
plt.plot(months, jac_by_month, marker="o")
plt.xticks(months)
plt.xlabel("Month")
plt.ylabel("Jaccard(spring, month)")
plt.title("How similar each month’s species are to spring")
plt.tight_layout()
plt.show()

plt.figure()
plt.bar(months, n_by_month)
plt.xticks(months)
plt.xlabel("Month")
plt.ylabel("# species in month")
plt.title("Species richness by month (for context)")
plt.tight_layout()
plt.show()
```
